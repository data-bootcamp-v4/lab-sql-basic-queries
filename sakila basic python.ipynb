```python
# LAB | SQL Basic Queries - Sakila Database
# Python/Jupyter Notebook Version

import pandas as pd
import mysql.connector
# Alternative imports:
# import sqlite3  # For SQLite
# import psycopg2  # For PostgreSQL

# Database connection configuration
DB_CONFIG = {
    'host': 'localhost',
    'user': 'your_username',
    'password': 'your_password',
    'database': 'sakila'
}

def execute_query(query, description, show_results=True):
    """Execute a SQL query and return results"""
    try:
        conn = mysql.connector.connect(**DB_CONFIG)
        
        if query.strip().upper().startswith('SHOW'):
            cursor = conn.cursor()
            cursor.execute(query)
            results = cursor.fetchall()
            columns = [desc[0] for desc in cursor.description]
            df = pd.DataFrame(results, columns=columns)
            cursor.close()
        else:
            df = pd.read_sql(query, conn)
        
        conn.close()
        
        if show_results:
            print(f"\n{'='*60}")
            print(f"QUERY: {description}")
            print(f"{'='*60}")
            print(f"SQL: {query}")
            print(f"\nRESULTS:")
            if len(df) > 20:
                print(f"Showing first 20 rows of {len(df)} total rows:")
                print(df.head(20))
            else:
                print(df)
            print(f"Total rows: {len(df)}")
        
        return df
        
    except Exception as e:
        print(f"Error executing query: {e}")
        return None

# =====================================================
# 1. Display all available tables
# =====================================================

query1 = "SHOW TABLES"
result1 = execute_query(query1, "1. All tables in Sakila database")

# =====================================================
# 2. Retrieve all data from actor, film, customer tables
# =====================================================

query2a = "SELECT * FROM actor"
result2a = execute_query(query2a, "2A. All data from actor table")

query2b = "SELECT * FROM film"  
result2b = execute_query(query2b, "2B. All data from film table")

query2c = "SELECT * FROM customer"
result2c = execute_query(query2c, "2C. All data from customer table")

# =====================================================
# 3. Retrieve specific columns
# =====================================================

query3_1 = "SELECT title FROM film"
result3_1 = execute_query(query3_1, "3.1. All film titles")

query3_2 = "SELECT name AS language FROM language"
result3_2 = execute_query(query3_2, "3.2. All languages (aliased)")

query3_3 = "SELECT first_name FROM staff"
result3_3 = execute_query(query3_3, "3.3. Staff first names")

# =====================================================
# 4. Unique release years
# =====================================================

query4 = "SELECT DISTINCT release_year FROM film"
result4 = execute_query(query4, "4. Unique release years")

# =====================================================
# 5. Counting records
# =====================================================

query5_1 = "SELECT COUNT(*) AS number_of_stores FROM store"
result5_1 = execute_query(query5_1, "5.1. Number of stores")

query5_2 = "SELECT COUNT(*) AS number_of_employees FROM staff"
result5_2 = execute_query(query5_2, "5.2. Number of employees")

query5_3a = "SELECT COUNT(*) AS films_available_for_rent FROM inventory"
result5_3a = execute_query(query5_3a, "5.3A. Films available for rent")

query5_3b = "SELECT COUNT(DISTINCT inventory_id) AS films_rented FROM rental"
result5_3b = execute_query(query5_3b, "5.3B. Films that have been rented")

query5_4 = "SELECT COUNT(DISTINCT last_name) AS distinct_last_names FROM actor"
result5_4 = execute_query(query5_4, "5.4. Distinct actor last names")

# =====================================================
# 6. 10 longest films
# =====================================================

query6 = """
SELECT title, length 
FROM film 
ORDER BY length DESC 
LIMIT 10
"""
result6 = execute_query(query6, "6. 10 longest films")

# =====================================================
# 7. Filtering techniques
# =====================================================

query7_1 = "SELECT * FROM actor WHERE first_name = 'SCARLETT'"
result7_1 = execute_query(query7_1, "7.1. Actors named SCARLETT")

# =====================================================
# BONUS QUESTIONS
# =====================================================

bonus7_2 = """
SELECT title, length 
FROM film 
WHERE title LIKE '%ARMAGEDDON%' 
AND length > 100
"""
result_bonus7_2 = execute_query(bonus7_2, "BONUS 7.2. ARMAGEDDON movies > 100 minutes")

bonus7_3 = """
SELECT COUNT(*) AS films_with_behind_scenes 
FROM film 
WHERE special_features LIKE '%Behind the Scenes%'
"""
result_bonus7_3 = execute_query(bonus7_3, "BONUS 7.3. Films with Behind the Scenes")

# =====================================================
# SUMMARY REPORT
# =====================================================

print("\n" + "="*80)
print("SAKILA DATABASE SUMMARY REPORT")
print("="*80)

if result1 is not None and len(result1) > 0:
    print(f"Total tables in database: {len(result1)}")

if result2a is not None:
    print(f"Total actors: {len(result2a)}")

if result2b is not None:
    print(f"Total films: {len(result2b)}")
    
if result2c is not None:
    print(f"Total customers: {len(result2c)}")

if result4 is not None:
    years = result4['release_year'].tolist()
    print(f"Release years: {years}")

if result5_1 is not None:
    print(f"Number of stores: {result5_1['number_of_stores'].iloc[0]}")

if result5_2 is not None:
    print(f"Number of employees: {result5_2['number_of_employees'].iloc[0]}")

if result5_3a is not None and result5_3b is not None:
    available = result5_3a['films_available_for_rent'].iloc[0]
    rented = result5_3b['films_rented'].iloc[0]
    print(f"Films available for rent: {available}")
    print(f"Films that have been rented: {rented}")

if result7_1 is not None:
    print(f"Actors named SCARLETT: {len(result7_1)}")

if result_bonus7_2 is not None:
    print(f"ARMAGEDDON movies > 100 min: {len(result_bonus7_2)}")

if result_bonus7_3 is not None:
    print(f"Films with Behind the Scenes: {result_bonus7_3['films_with_behind_scenes'].iloc[0]}")

print("="*80)
print("All queries completed successfully!")
print("Remember to update DB_CONFIG with your database credentials.")
```
