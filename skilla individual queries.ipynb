```python
-- =====================================================
-- LAB | SQL Basic Queries - Individual Query Version
-- Copy and paste each query separately into your SQL client
-- =====================================================

-- QUERY 1: Show all tables
SHOW TABLES;

-- =====================================================

-- QUERY 2A: All data from actor table
SELECT * FROM actor;

-- =====================================================

-- QUERY 2B: All data from film table  
SELECT * FROM film;

-- =====================================================

-- QUERY 2C: All data from customer table
SELECT * FROM customer;

-- =====================================================

-- QUERY 3.1: Titles of all films
SELECT title FROM film;

-- =====================================================

-- QUERY 3.2: Languages with alias
SELECT name AS language FROM language;

-- =====================================================

-- QUERY 3.3: Staff first names
SELECT first_name FROM staff;

-- =====================================================

-- QUERY 4: Unique release years
SELECT DISTINCT release_year FROM film;

-- =====================================================

-- QUERY 5.1: Number of stores
SELECT COUNT(*) AS number_of_stores FROM store;

-- =====================================================

-- QUERY 5.2: Number of employees
SELECT COUNT(*) AS number_of_employees FROM staff;

-- =====================================================

-- QUERY 5.3A: Films available for rent
SELECT COUNT(*) AS films_available_for_rent FROM inventory;

-- =====================================================

-- QUERY 5.3B: Films that have been rented
SELECT COUNT(DISTINCT inventory_id) AS films_rented FROM rental;

-- =====================================================

-- QUERY 5.4: Distinct actor last names
SELECT COUNT(DISTINCT last_name) AS distinct_last_names FROM actor;

-- =====================================================

-- QUERY 6: 10 longest films
SELECT title, length 
FROM film 
ORDER BY length DESC 
LIMIT 10;

-- =====================================================

-- QUERY 7.1: Actors named SCARLETT
SELECT * FROM actor WHERE first_name = 'SCARLETT';

-- =====================================================

-- BONUS 7.2: ARMAGEDDON movies longer than 100 minutes
SELECT title, length 
FROM film 
WHERE title LIKE '%ARMAGEDDON%' 
AND length > 100;

-- =====================================================

-- BONUS 7.3: Films with Behind the Scenes content
SELECT COUNT(*) AS films_with_behind_scenes 
FROM film 
WHERE special_features LIKE '%Behind the Scenes%';
```
